```python
import itertools
import pandas as pd
import matplotlib.pyplot as plt

```

```python
hyperparameter_config = {
    "learning_rate": [0.001, 0.0005, 0.0001],
    "batch_size": [32, 64],
    "gamma": [0.95, 0.99],
    "epsilon_decay": [0.995, 0.999],
    "target_update": [10, 20]
}

hyperparameter_config
```

```python
keys = hyperparameter_config.keys()

values = hyperparameter_config.values()

experiments = list(itertools.product(*values))

print("Total Experiments :", len(experiments))

experiments[:5]
```

```python
df = pd.DataFrame(
    experiments,
    columns=keys
)

df.head()
```

```python
df["Average Reward"] = None
df["Loss"] = None
df["Training Time"] = None

df.head()
```

```python
df.to_csv(
    "hyperparameter_experiments.csv",
    index=False
)

print("Experiment sheet created successfully.")
```

```python
def plot_results(df):

    if df["Average Reward"].isnull().all():

        print("No results available.")

        return

    plt.figure(figsize=(10,5))

    plt.plot(df["Average Reward"])

    plt.xlabel("Experiment")

    plt.ylabel("Average Reward")

    plt.title("Hyperparameter Comparison")

    plt.grid(True)

    plt.show()

    plot_results(df)
```

```python
def best_configuration(df):

    if df["Average Reward"].isnull().all():

        print("Training results not available.")

        return

    best = df.loc[df["Average Reward"].idxmax()]

    print(best)
```
