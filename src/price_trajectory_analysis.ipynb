```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Temporary data
# Replace with actual DQN output after team merge

episodes = np.arange(1, 101)

prices = np.random.randint(2500, 7000, 100)

rewards = np.random.uniform(100, 300, 100)

revenues = prices * np.random.randint(15, 40, 100)

baseline_revenue = np.full(100, 90000)

occupancy = np.random.uniform(60, 100, 100)

results = pd.DataFrame({
    "Episode": episodes,
    "Price": prices,
    "Reward": rewards,
    "Revenue": revenues,
    "Baseline Revenue": baseline_revenue,
    "Occupancy": occupancy
})

results.head()
```

```python
plt.figure(figsize=(10,5))

plt.plot(results["Episode"], results["Price"])

plt.title("Price Trajectory")

plt.xlabel("Episode")

plt.ylabel("Price")

plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(results["Episode"], results["Reward"])

plt.title("Reward per Episode")

plt.xlabel("Episode")

plt.ylabel("Reward")

plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(results["Episode"], results["Revenue"], label="DQN")

plt.plot(results["Episode"], results["Baseline Revenue"], label="Baseline")

plt.legend()

plt.title("Revenue Comparison")

plt.xlabel("Episode")

plt.ylabel("Revenue")

plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(results["Episode"], results["Occupancy"])

plt.title("Occupancy Trend")

plt.xlabel("Episode")

plt.ylabel("Occupancy (%)")

plt.grid()

plt.show()
```

```python
summary = pd.DataFrame({

    "Metric":[
        "Average Price",
        "Average Reward",
        "Average Revenue",
        "Average Occupancy"
    ],

    "Value":[
        results["Price"].mean(),
        results["Reward"].mean(),
        results["Revenue"].mean(),
        results["Occupancy"].mean()
    ]

})

summary
```

```python
summary.to_csv("performance_summary.csv", index=False)

print("Performance summary saved successfully.")
```
